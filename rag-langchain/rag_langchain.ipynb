```python
from pinecone import Pinecone, ServerlessSpec
from google.colab import userdata
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_pinecone import PineconeVectorStore
```

```python
PINECONE_API_KEY = userdata.get('PINECONE_API_KEY')
pc = Pinecone(api_key=PINECONE_API_KEY)
```

```python
index_name = "rag-langchain"

if index_name not in [i["name"] for i in pc.list_indexes()]:

    pc.create_index(
        name=index_name,
        dimension=384,
        metric="cosine",
        spec=ServerlessSpec(
            cloud="aws",
            region="us-east-1"
        )
    )
```

```python
index = pc.Index(index_name)
```

```python
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
documents = [
    "LangChain is an open-source framework for building applications using Large Language Models.",
    
    "Pinecone is a vector database used to store and retrieve embeddings efficiently.",
    
    "Retrieval Augmented Generation, also known as RAG, combines retrieval and generation techniques.",
    
    "Flask is a lightweight Python web framework used for building APIs and web applications.",
    
    "Embeddings are vector representations of text used for semantic search and similarity matching.",
    
    "LangChain provides chains, agents, memory, retrievers, prompt templates, and integrations with LLMs.",
    
    "Vector databases help perform semantic search using high-dimensional vectors.",
    
    "Gemini is Google's family of generative AI models."
]
```

```python
vectorstore = PineconeVectorStore.from_texts(
    texts=documents,
    embedding=embeddings,
    index_name=index_name,
    pinecone_api_key=PINECONE_API_KEY
)
```
